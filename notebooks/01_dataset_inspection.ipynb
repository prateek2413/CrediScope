```python

from collections import Counter
import pandas as pd

status_counts = Counter()

chunksize = 250_000

for chunk in pd.read_csv(
    file_path,
    usecols=["loan_status"],
    chunksize=chunksize
):
    status_counts.update(
        chunk["loan_status"].value_counts(dropna=False).to_dict()
    )

status_distribution = pd.Series(status_counts).sort_values(
    ascending=False
)

print("FULL DATASET LOAN STATUS DISTRIBUTION")
display(
    status_distribution.to_frame(name="Count")
)

print("\nTotal records:", sum(status_counts.values()))
```

```python

fully_paid = status_counts.get("Fully Paid", 0)

defaulted = (
    status_counts.get("Charged Off", 0)
    + status_counts.get("Default", 0)
)

eligible_total = fully_paid + defaulted

target_distribution = pd.DataFrame({
    "Loan Outcome": ["Non-Default", "Default"],
    "Count": [fully_paid, defaulted],
    "Percentage": [
        fully_paid / eligible_total * 100,
        defaulted / eligible_total * 100
    ]
})

display(target_distribution)

print("Eligible records:", eligible_total)
print("Excluded records:", sum(status_counts.values()) - eligible_total)
```

```python

import pandas as pd

selected_features = [
    "loan_amnt",
    "term",
    "int_rate",
    "installment",
    "grade",
    "emp_length",
    "home_ownership",
    "annual_inc",
    "verification_status",
    "issue_d",
    "purpose",
    "dti",
    "delinq_2yrs",
    "inq_last_6mths",
    "open_acc",
    "revol_bal",
    "revol_util",
    "total_acc",
    "pub_rec",
    "loan_status"
]

status_mapping = {
    "Fully Paid": 0,
    "Charged Off": 1,
    "Default": 1
}

missing_counts = pd.Series(0, index=selected_features, dtype="int64")
eligible_count = 0

for chunk in pd.read_csv(
    file_path,
    usecols=selected_features,
    chunksize=100_000,
    low_memory=False
):
    chunk = chunk[chunk["loan_status"].isin(status_mapping)]

    eligible_count += len(chunk)

    missing_counts += chunk.isna().sum()

missing_analysis = pd.DataFrame({
    "Missing Count": missing_counts,
    "Missing Percentage": (
        missing_counts / eligible_count * 100
    ).round(2)
})

display(
    missing_analysis.sort_values(
        "Missing Percentage",
        ascending=False
    )
)

print("Eligible records processed:", eligible_count)
```
